```python
import pandas as pd
from IPython.display import display

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 100)


df = pd.read_csv("Titanic_Survival_Dataset.csv")

print("========= FIRST 5 ROWS =========")
display(df.head())

print("\n========= LAST 5 ROWS =========")
display(df.tail())

print("\n========= SHAPE =========")
print(df.shape)

print("\n========= COLUMN NAMES =========")
print(df.columns.tolist())

print("\n========= DATA TYPES =========")
print(df.dtypes)

print("\n========= MISSING VALUES =========")
print(df.isnull().sum())

print("\n========= STATISTICAL SUMMARY =========")
display(df.describe())

print("\n========= SURVIVAL COUNT =========")
print(df["Survived"].value_counts())

print("\n========= GENDER COUNT =========")
print(df["Sex"].value_counts())

print("\n========= PASSENGER CLASS COUNT =========")
print(df["Pclass"].value_counts())

print("\n========= AVERAGE AGE =========")
print(df["Age"].mean())

print("\n========= HIGHEST FARE =========")
print(df["Fare"].max())

print("\n========= OLDEST PASSENGER =========")
display(df[df["Age"] == df["Age"].max()])

print("\n========= Youngest PASSENGER =========")
display(df[df["Age"] == df["Age"].min()])

print("\n========= FEMALE PASSENGERS =========")
display(df[df["Sex"] == "female"].head())

print("\n========= PASSENGERS WHO SURVIVED =========")
display(df[df["Survived"] == 1].head())

print("\n========= SORT BY FARE (DESCENDING) =========")
display(df.sort_values(by="Fare", ascending=False).head(10))

print("\n========= Survival Rate) =========")
survivalrate = df["Survived"].mean() * 100
print(f"Survival Rate: {survivalrate:.2f}%")

print("\n========= Survival Rate Acc. to Passenger Class =========")
print(df.groupby("Pclass")["Survived"].mean())

print("\n========= Average Fare Paid =========")
print(df["Fare"].mean())

print("\n========= Unique Pclass  =========")
print(df["Pclass"].nunique())

print("\n========= Unique Tickets Types  =========")
print(df["Ticket"].nunique())

print("\n========= Passenger Name Start with B  =========")

display(df[df["Name"].str.startswith("B")])
```
