```python
# Product Catalog
 
products = {
    "P101": {"name": "Laptop", "price": 50000, "stock": 10},
    "P102": {"name": "Mouse", "price": 500, "stock": 50},
    "P103": {"name": "Keyboard", "price": 1500, "stock": 0}
}
 
# 1. Add Product
products["P104"] = {"name": "Monitor", "price": 12000, "stock": 5}
print("After Adding Product:")
print(products)
 
# 2. Search by Product ID
pid = "P102"
if pid in products:
    print("\nProduct Found:")
    print(products[pid])
else:
    print("Product not found")
 
# 3. Update Stock
products["P101"]["stock"] = 15
print("\nUpdated Stock of Laptop:")
print(products["P101"])
 
# 4. Find Most Expensive Product
expensive = max(products, key=lambda x: products[x]["price"])
 
print("\nMost Expensive Product:")
print("ID:", expensive)
print("Name:", products[expensive]["name"])
print("Price:", products[expensive]["price"])
 
# 5. Calculate Total Inventory Value
total_value = 0
 
for pid, details in products.items():
    total_value += details["price"] * details["stock"]
 
print("\nTotal Inventory Value = ₹", total_value)
 
# 6. Display Out-of-Stock Products
print("\nOut-of-Stock Products:")
for pid, details in products.items():
    if details["stock"] == 0:
        print(pid, "-", details["name"])
```
