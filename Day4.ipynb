```python
import pandas as pd

data = {
    "Hero": [
        " iron man ",
        "THOR",
        "hulk",
        "Captain America",
        "Spider-Man123"
    ]
}

df = pd.DataFrame(data)

df["strip"] = df["Hero"].str.strip()
df["upper"] = df["Hero"].str.upper()
df["lower"] = df["Hero"].str.lower()
df["capitalize"] = df["Hero"].str.capitalize()
df["title"] = df["Hero"].str.title()
df["length"] = df["Hero"].str.len()
df["replace"] = df["Hero"].str.replace("man", "hero", case=False)
df["starts_I"] = df["Hero"].str.startswith("I")
df["contains_man"] = df["Hero"].str.contains("man", case=False)
df["count_a"] = df["Hero"].str.count("a")
df["find_a"] = df["Hero"].str.find("a")
df["first_char"] = df["Hero"].str[0]
df["last_char"] = df["Hero"].str[-1]
df["split_words"] = df["Hero"].str.split()
df["is_alpha"] = df["Hero"].str.isalpha()
df["is_alnum"] = df["Hero"].str.isalnum()

print(df)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Amit', 'Pooja', 'Ravi', 'Neha']
})

df2 = pd.DataFrame({
    'ID': [3, 4, 5, 6],
    'Subject': ['Math', 'Science', 'English', 'History']
})

inner_join = pd.merge(df1, df2, on="ID", how="inner")
outer_join = pd.merge(df1, df2, on="ID", how="outer")
left_join = pd.merge(df1, df2, on="ID", how="left")
right_join = pd.merge(df1, df2, on="ID", how="right")
cross_join = pd.merge(df1, df2, how="cross")

print("===== INNER JOIN =====")
print(inner_join)

print("\n===== OUTER JOIN =====")
print(outer_join)

print("\n===== LEFT JOIN =====")
print(left_join)

print("\n===== RIGHT JOIN =====")
print(right_join)

print("\n===== CROSS JOIN =====")
print(cross_join)
```
