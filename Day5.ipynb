```python
import pandas as pd

df = pd.read_csv("Titanic_Survival_Dataset.csv")

print(df.head())
print(df.columns)
```

```python
import matplotlib.pyplot as plt

pclass_count = df["Pclass"].value_counts()

plt.bar(pclass_count.index, pclass_count.values)
plt.title("Passengers by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
gender_count = df["Sex"].value_counts()

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")
plt.show()
```

```python
plt.hist(df["Age"].dropna(), bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Passengers")
plt.show()
```

```python
survival = df["Survived"].value_counts()

plt.bar(
    ["Not Survived", "Survived"],
    survival.values,
    color=["red", "green"]
)

plt.title("Survival Report")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Paper_leaks_India.csv", encoding="cp1252")

exam_count = df["Exam Name"].value_counts().head(10)

plt.figure(figsize=(10,5))
plt.bar(exam_count.index, exam_count.values, color="red")

plt.title("Top 10 Exams with Paper Leaks")
plt.xlabel("Exam Name")
plt.ylabel("Number of Incidents")

plt.xticks(rotation=90)

plt.show()
```

```python
areas = df["Area(s) of Incident"].value_counts().head(10)

plt.barh(
    areas.index,
    areas.values,
    color="green"
)

plt.title("Top Areas of Incidents")
plt.xlabel("Count")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

sns.countplot(
    y="Conducting Body Type",
    data=df,
    order=df["Conducting Body Type"].value_counts().index
)

plt.title("Conducting Body Type Count")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Paper_leaks_India.csv", encoding="cp1252")

top_exams = df["Exam Name"].value_counts().head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top_exams.index,
    top_exams.values,
    color="crimson"
)

plt.title("Top 10 Exams Affected by Paper Leaks")
plt.xlabel("Exam")
plt.ylabel("Leak Count")

plt.xticks(rotation=90)

for i, v in enumerate(top_exams.values):
    plt.text(i, v, str(v), ha="center")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Paper_leaks_India.csv", encoding="cp1252")

plt.figure(figsize=(12,6))

sns.countplot(
    y="Exam Name",
    data=df,
    order=df["Exam Name"].value_counts().head(10).index
)

plt.title("Top 10 Exams with Paper Leak Incidents")
plt.show()
```

```python
df["Date of Exam/Incident"] = pd.to_datetime(
    df["Date of Exam/Incident"],
    errors="coerce"
)

df["Year"] = df["Date of Exam/Incident"].dt.year

year_count = df["Year"].value_counts().sort_index()

sns.lineplot(
    x=year_count.index,
    y=year_count.values,
    marker="o"
)

plt.title("Paper Leaks by Year")
plt.xlabel("Year")
plt.ylabel("Incidents")

plt.show()
```

```python
import seaborn as sns

numeric_cols = df.select_dtypes(include="number")

sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Paper_leaks_India.csv", encoding="cp1252")


df["Appeared Students"] = pd.to_numeric(
    df["Appeared Students"],
    errors="coerce"
)


df["Date of Exam/Incident"] = pd.to_datetime(
    df["Date of Exam/Incident"],
    errors="coerce"
)

df["Year"] = df["Date of Exam/Incident"].dt.year


numeric_df = df.select_dtypes(include="number")


corr = numeric_df.corr()


plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```
