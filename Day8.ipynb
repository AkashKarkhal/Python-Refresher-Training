```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["TV", "Fridge", "AC"],
    "Price": [20000, 25000, 35000],
    "Cost": [15000, 18000, 28000]
})

df["New_Price"] = round(df["Price"] * 1.08, 2)

df["Profit_Margin_%"] = round(((df["Price"] - df["Cost"]) / df["Price"]) * 100, 2)

print(df)
```
