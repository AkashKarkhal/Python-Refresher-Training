```python
from datetime import datetime, timedelta


class CoffeeOrder:
    def __init__(self, customer_name, coffee_type, size, quantity):
        self.customer_name = customer_name
        self.coffee_type = coffee_type
        self.size = size
        self.quantity = quantity

        self.order_time = datetime.now()
        self.delivery_time = self.order_time + timedelta(minutes=15)

    def bill_amount(self):
        prices = {
            "small": 100,
            "medium": 150,
            "large": 200
        }
        return prices.get(self.size.lower(), 150) * self.quantity

    def show_order_summary(self):
        print("\n===== Coffee Order Summary =====")
        print(f"Customer Name      : {self.customer_name}")
        print(f"Coffee Type        : {self.coffee_type}")
        print(f"Size               : {self.size}")
        print(f"Quantity           : {self.quantity}")
        print(f"Order Time         : {self.order_time.strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"Expected Delivery  : {self.delivery_time.strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"Total Bill         : ₹{self.bill_amount()}")
        print("================================")


# User Input
name = input("Enter your name: ")
coffee = input("Enter coffee type (Latte/Espresso/Cappuccino): ")
size = input("Enter size (Small/Medium/Large): ")
qty = int(input("Enter quantity: "))

# Create Order
order1 = CoffeeOrder(name, coffee, size, qty)

# Display Summary
order1.show_order_summary()
```

```python
def generate_bill(customer_name,*items):
    print(f"Bill for {customer_name}:")
    print("-----------------------")
    print("                      ")
    print("-----------------------")
    total=0
    for item in items:
        print(f"{item[0]}:${item[1]}")
        total+=item[1]
    print(f"Total Amount: ${total}\n")
    print("------------------------")
generate_bill("Riya",("Shampoo",200),("Soap",120),("Facewash",300))  
```

```python
def bill(name, **det):
    print(f"Customer Name: {name}")
    print("------ Bill ------")
    total = 0
    for food, details in det.items():
        price = details["price"]
        quantity = details["quantity"]
        cost = price * quantity
        print(f"{food} - {quantity} x {price} = {cost}")
        total += cost
    print("------------------")
    print(f"Total Bill : {total}")
    print("------------------")

 
 
bill("Aman",
     Pizza={"price": 100, "quantity": 3},
     Burger={"price": 80, "quantity": 2},
     Coke={"price": 50, "quantity": 1})
```

```python
def gym_bill(member_name, **plans):
    print(f"Member Name: {member_name}")
    print("----- Gym Bill -----")

    total = 0

    for service, details in plans.items():
        price = details["price"]
        months = details["months"]

        cost = price * months

        print(f"{service} - {months} month(s) x {price} = {cost}")
        total += cost

    print("--------------------")
    print(f"Total Amount: {total}")
    print("--------------------")


gym_bill(
    "Akash",
    GymMembership={"price": 1500, "months": 3},
    PersonalTraining={"price": 2000, "months": 2},
    DietPlan={"price": 500, "months": 1}
)
```

```python
class Workout:

    def __init__(self, steps):
        self.steps = steps

    def show_calories(self):
        calories = self.steps * 0.04
        print(f"Steps: {self.steps}")
        print(f"Calories Burned: {calories}")


w1 = Workout(2000)
w2 = Workout(5000)
w3 = Workout(10000)

w1.show_calories()
print()

w2.show_calories()
print()

w3.show_calories()
```

```python
class Vehicle:
    def start_engine(self):
        print("Engine is started")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class Bike(Vehicle):
    def ride(self):
        print("Bike is riding")


car1 = Car()
bike1 = Bike()

car1.start_engine()   # Inherited from Vehicle
car1.drive()

print()

bike1.start_engine()  # Inherited from Vehicle
bike1.ride()
```

```python
class FitnessTracker:

    def track_activity(self):
        print("Tracking fitness activity...")


class Workout(FitnessTracker):

    def show_workout(self):
        print("Workout session started")


class CardioWorkout(Workout):

    def calories_burned(self, steps):
        calories = steps * 0.04
        print(f"You burned {calories} calories")


# Object of grandchild class
c1 = CardioWorkout()

c1.track_activity()      # From FitnessTracker
c1.show_workout()        # From Workout
c1.calories_burned(5000) # From CardioWorkout
```
