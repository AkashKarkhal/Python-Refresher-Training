```python
import numpy as np
temperature_data=np.random.randint(20,45,size=(30,4))
print(temperature_data)
daily_avg=np.mean(temperature_data,axis=1)
print("Daily Average Temperature",daily_avg)
first_28_days=temperature_data[:28]
weekly_avg=np.mean(first_28_days.reshape(4,7,4),axis=(1,2))
print("weekly average based on first 4 full weeks):\n",weekly_avg)
 
hottest_day=np.argmax(daily_avg)
coldest_day=np.argmin(daily_avg)
print(f"Hottest day: Day{hottest_day+1} with {daily_avg[hottest_day]:.2f}C")
print(f"Coldest day: Day{coldest_day+1} with {daily_avg[coldest_day]:.2f}C")
```

```python
import numpy as np

suspects = np.array(["John", "Mike", "Alex", "David"])

evidence = np.array([
    [85, 90, 75, 80],
    [70, 65, 80, 85],
    [95, 92, 90, 94],
    [60, 75, 55, 70]
])

print("Evidence Matrix")
print(evidence)

total_score = np.sum(evidence, axis=1)

print("\nTotal Evidence Score")
for name, score in zip(suspects, total_score):
    print(name, ":", score)

main_suspect = suspects[np.argmax(total_score)]

print("\nMost Likely Criminal:", main_suspect)
print("Highest Score:", np.max(total_score))
```
